## 1. What are we using, and why?

**Computational physics** means using a computer to do physics that would be hard or impossible with pencil and paper alone. There are three big flavors:

1. **Simulation**: you write down the laws of physics (for example Newton's second law or the Schrödinger equation) and let the computer step them forward to see what happens.
2. **Numerical methods**: you get approximate numbers for things you cannot do analytically, such as an integral with no closed form or the energy levels of a strange potential well.
3. **Data analysis**: you take measurements and extract physics from them, for example fitting a curve to find a lifetime or a mass.

To do any of this we need a way to give instructions to the computer. In this course we use:

- **Python**: a *programming language*. A program is a list of instructions written in a precise, unambiguous way. Python is popular in physics because it is readable and has excellent free libraries for numbers, plots, and data.
- **Jupyter notebooks**: documents (files ending in `.ipynb`) that mix *code you can run*, the *output* of that code (numbers, plots), and *formatted text and equations*. You are reading one right now.
- **Google Colab**: a free website from Google that runs Jupyter notebooks for you "in the cloud". The Python code does not run on your laptop; it runs on a Google computer, and you see the results in your browser. Nothing needs to be installed.

You need a Google account to use Colab. A personal Gmail account is fine (your OU account is a Microsoft account, not a Google one).

## 4. Variables and the order of execution

You can store a value under a name using `=`. This is called **assignment**, and the name is called a **variable**. (The next notebook covers variables carefully; here we only need the basic idea.)

````python
speed_of_light = 299792458   # in m/s (exact, by definition of the meter)
print(speed_of_light)
````

Once a variable is defined, **any later cell can use it**, because all the cells in a notebook share the same Python "memory". This memory is called the **kernel** (or in Colab, the **session** or **runtime**).

````python
# Distance light travels in one nanosecond (1 ns = 1e-9 s), in meters
speed_of_light * 1e-9
````

Light travels about 30 cm in a nanosecond. (`1e-9` is how you write $1\times10^{-9}$ in Python.)

### The hidden-state pitfall

Here is the single most common source of confusion in notebooks: **what matters is the order in which you *ran* the cells, not the order in which they appear on the page.**

Look at the number in square brackets to the left of each code cell you have run, like `[3]`. That is the *execution count*: it tells you the order in which cells were run. If the numbers are not increasing from top to bottom, you ran things out of order.

Try this experiment:

1. Run the next cell once. It sets `counter` to 0.
2. Run the cell after it **several times** in a row (click in it and press **Ctrl+Enter** repeatedly).

````python
counter = 0
````

````python
counter = counter + 1   # take the current value, add 1, and store it back
print(counter)
````

Each time you run the second cell, `counter` goes up by one: 1, 2, 3, ... The page looks exactly the same each time, but the hidden value in memory keeps changing. Someone reading your notebook would expect to see `1`.

Other ways hidden state bites:

- You define a variable, then **delete or edit** the cell that defined it. The old value is still in memory, so everything seems to work... until you (or your grader) open the notebook fresh, and it crashes with `NameError: name '...' is not defined`.
- You change a value in an early cell but forget to rerun the later cells that depend on it, so your plots and numbers are out of date.

### The fix: Restart and run all

Colab's **Runtime** menu has the tools you need:

- **Runtime → Restart session**: wipes Python's memory clean (your code and text are *not* deleted, only the variables).
- **Runtime → Run all**: runs every cell from top to bottom.
- **Runtime → Restart session and run all**: both at once. This is the honest test of your notebook. If it runs from top to bottom without errors and gives the answers you expect, your notebook is in good shape.

**Always do Restart session and run all before you submit anything.** (The exact menu wording changes slightly from time to time; look for "restart" and "run all".)

### Runtimes disconnect

If you leave Colab idle for a while (roughly an hour or two, sometimes less), Google disconnects your runtime to free the computer for others. Your notebook text and code are safe in Drive, but **all variables are gone**. When you come back, use **Runtime → Run all** (or run the cells you need from the top) before continuing. If you see `NameError` for something you are sure you defined, this is probably why.

## 6. Writing equations with LaTeX

Physics needs equations, and Markdown cells understand **LaTeX** (pronounced "lay-tech"), the standard language physicists use to typeset math. Double-click this cell to see how the equations below are written.

- Put math **inside a sentence** between single dollar signs: `$E = \gamma m c^2$` gives $E = \gamma m c^2$.
- Put math **on its own centered line** between double dollar signs:

```
$$\gamma = \frac{1}{\sqrt{1 - v^2/c^2}}$$
```

gives

$$\gamma = \frac{1}{\sqrt{1 - v^2/c^2}}$$

The most common LaTeX pieces:

| You type | You get |
|---|---|
| `x^2`, `e^{-x/\lambda}` | $x^2$, $e^{-x/\lambda}$ (use braces `{}` when the exponent is more than one character) |
| `x_0`, `E_{n}`, `p_{\rm max}` | $x_0$, $E_{n}$, $p_{\rm max}$ (subscripts) |
| `\frac{a}{b}` | $\frac{a}{b}$ |
| `\sqrt{x}` | $\sqrt{x}$ |
| `\alpha, \beta, \gamma, \lambda, \psi, \hbar` | $\alpha, \beta, \gamma, \lambda, \psi, \hbar$ |
| `\Delta, \Psi, \Omega` | $\Delta, \Psi, \Omega$ (capital Greek letters) |
| `\int_0^\infty f(x)\,dx` | $\int_0^\infty f(x)\,dx$ |
| `\frac{\partial \psi}{\partial t}` | $\frac{\partial \psi}{\partial t}$ |
| `\approx, \times, \cdot, \pm, \infty` | $\approx, \times, \cdot, \pm, \infty$ |

### Some equations from Modern Physics

The photon energy and the de Broglie wavelength:
$$E = h f = \frac{hc}{\lambda}, \qquad \lambda = \frac{h}{p}.$$

The relativistic energy-momentum relation:
$$E^2 = (pc)^2 + (mc^2)^2.$$

The energy levels of hydrogen:
$$E_n = -\frac{13.6\ \text{eV}}{n^2}, \qquad n = 1, 2, 3, \ldots$$

The time-dependent Schrödinger equation in one dimension:
$$i\hbar \frac{\partial \Psi(x,t)}{\partial t} = -\frac{\hbar^2}{2m}\frac{\partial^2 \Psi(x,t)}{\partial x^2} + V(x)\,\Psi(x,t).$$

and the time-independent version:
$$-\frac{\hbar^2}{2m}\frac{d^2\psi}{dx^2} + V(x)\,\psi(x) = E\,\psi(x).$$

Double-click this cell and study how each one is written. A few details worth noticing: `\qquad` adds a wide space, `\ ` (backslash-space) adds a small space, and `\text{eV}` writes upright text inside math.

> **Pitfall:** In LaTeX, every `{` needs a matching `}` and every `$` needs a matching `$`. If an equation shows up as raw code instead of formatted math, look for a missing brace or dollar sign.

## 8. A first taste of computational physics

Let us do something a little more interesting than $2+3$. Do not worry about understanding every line yet; the next few notebooks explain all of it. The goal is to see what a short calculation looks like.

**A bit of physics first.** A *muon* is an unstable particle, like a heavy cousin of the electron, that decays on average after a short time $\tau$ measured in its own rest frame. Special relativity says that a clock moving at speed $v$ runs slow by the *Lorentz factor* $\gamma$ as seen from the ground: a time $\tau$ for the muon lasts $\gamma\tau$ for us. This is *time dilation*. Since $\gamma \ge 1$, fast muons live longer in our frame.

**Question:** Cosmic-ray muons are produced high in the atmosphere and travel at nearly the speed of light. A muon at rest lives on average $\tau = 2.2\ \mu\text{s}$. If one moves at $v = 0.995c$, how far does it travel in one lifetime, (a) ignoring relativity and (b) including time dilation, where its lifetime in our frame is $\gamma\tau$ with
$$\gamma = \frac{1}{\sqrt{1 - \beta^2}}, \qquad \beta = v/c ?$$

````python
c = 299792458          # speed of light, m/s
tau = 2.2e-6           # muon lifetime at rest, s
beta = 0.995           # speed as a fraction of c

gamma = 1 / (1 - beta**2) ** 0.5      # ** means "to the power of"; power 0.5 is a square root
d_classical = beta * c * tau          # distance = speed * time, in m
d_relativistic = beta * c * gamma * tau

print("gamma =", gamma)
print("Distance without time dilation (m):", d_classical)
print("Distance with time dilation (m):", d_relativistic)
````

Without relativity, a typical muon would only make it about 650 m. With time dilation ($\gamma \approx 10$) it goes about 6.6 km, which is why we detect so many muons at sea level.

### A tiny plot

Computers really shine when you want to repeat a calculation many times. Here is $\gamma$ for many values of $\beta$ between 0 and 0.99, drawn as a curve. We use two *libraries* (collections of ready-made tools): `numpy` for numbers and `matplotlib` for plots. Notebooks 03 to 05 cover them properly.

````python
import numpy as np
import matplotlib.pyplot as plt

beta = np.linspace(0, 0.99, 200)       # 200 evenly spaced values from 0 to 0.99
gamma = 1 / np.sqrt(1 - beta**2)       # gamma for every one of them at once

plt.plot(beta, gamma)
plt.xlabel(r"$\beta = v/c$")
plt.ylabel(r"Lorentz factor $\gamma$")
plt.title("Time dilation grows rapidly as v approaches c")
plt.show()
````

Notice that LaTeX works in plot labels, too.
