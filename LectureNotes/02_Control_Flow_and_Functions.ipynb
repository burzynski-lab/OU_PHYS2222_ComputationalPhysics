## 1. Lists

A **list** is an ordered collection of values, written with square brackets and separated by commas:

````python
# Masses of some particles, in MeV/c^2 (rounded)
masses = [0.511, 105.7, 938.3, 939.6]
names = ["electron", "muon", "proton", "neutron"]

print(masses)
print(names)
print(type(masses))
````

### Indexing: getting one element

Each element has a position called its **index**. **Python counts from 0**, not 1: the first element is at index 0, the second at index 1, and so on. You get an element by writing the index in square brackets after the list's name.

| index | 0 | 1 | 2 | 3 |
|---|---|---|---|---|
| `names` | `"electron"` | `"muon"` | `"proton"` | `"neutron"` |
| negative index | -4 | -3 | -2 | -1 |

**Negative indices** count from the end: `-1` is the last element, `-2` the second-to-last, and so on.

````python
print(names[0])      # first element
print(names[1])      # second element
print(names[-1])     # last element
print(f"The {names[2]} mass is {masses[2]} MeV/c^2")
````

The **length** of a list (the number of elements) is given by `len`. Since counting starts at 0, the last valid index is `len(list) - 1`. Asking for an index that does not exist, like `names[4]` here, stops your program with the error

```
IndexError: list index out of range
```

````python
print(len(names))
print(names[len(names) - 1])     # the same as names[-1]
````

### Changing and adding elements

Lists can be changed after you create them. Assign to an index to replace an element, and use `.append(value)` to add an element at the end. (`append` is a **method**: a function that belongs to a particular object and is called with a dot after the object's name.)

A very common pattern is to start with an **empty list** `[]` and append to it.

````python
masses[0] = 0.51099895        # replace with a more precise value
print(masses)

names.append("muon neutrino")
print(names, len(names))

results = []                  # an empty list
results.append(1.5)
results.append(2.7)
print(results)
````

### Slicing: getting a piece of a list

A **slice** `list[start:stop]` gives a new list with the elements from index `start` up to **but not including** index `stop`. Leaving out `start` means "from the beginning"; leaving out `stop` means "to the end". The "not including the stop" rule takes some getting used to, but it means `list[0:3]` has exactly 3 elements.

````python
levels = [1, 2, 3, 4, 5, 6, 7, 8]

print(levels[0:3])    # indices 0, 1, 2
print(levels[2:5])    # indices 2, 3, 4
print(levels[:3])     # first three
print(levels[5:])     # from index 5 to the end
print(levels[-2:])    # last two
print(levels[::2])    # every second element: start:stop:step
````

## 2. `for` loops

Suppose you want to print the mass of every particle. You could write one `print` per particle, but that does not scale to a thousand particles. A **`for` loop** repeats a block of code once for each element of a list:

````python
for name in names:
    print("Particle:", name)

print("Done!")
````

How to read this:

- `for name in names:` means "for each element of the list `names`, set the variable `name` to that element, then run the indented block". The name `name` is your choice; pick something meaningful.
- The line ends with a **colon** `:`.
- The block of code to repeat is **indented** (by 4 spaces; Colab inserts them when you press Enter after a colon, or use Tab). The indented lines are called the **body** of the loop.
- The loop ends where the indentation ends. `print("Done!")` is not indented, so it runs once, after the loop finishes.

**In Python, indentation is not decoration: it is part of the grammar.** It is how Python knows which lines belong to the loop. Inconsistent indentation causes an `IndentationError`.

### Accumulating a result

A common pattern is to start a variable at zero and add to it inside the loop. Here is the total mass of the four particles in our list:

````python
# start again from the original four particles
names = ["electron", "muon", "proton", "neutron"]
masses = [0.511, 105.7, 938.3, 939.6]   # MeV/c^2

total = 0.0
for m in masses:
    total += m          # same as total = total + m
print(f"Total mass: {total:.1f} MeV/c^2")
````

### `range`: looping over numbers

Very often you want to loop over a sequence of integers. `range` produces them:

- `range(5)` gives 0, 1, 2, 3, 4 (five numbers, starting at 0, **stopping before 5**);
- `range(2, 6)` gives 2, 3, 4, 5;
- `range(1, 10, 2)` gives 1, 3, 5, 7, 9 (the third number is the step).

Like slices, `range` stops *before* the stop value.

````python
for i in range(5):
    print(i, i**2)
````

````python
for n in range(1, 10, 2):
    print(n, end=" ")     # end=" " prints a space instead of starting a new line
````

To see all the numbers in a `range` at once, convert it to a list with `list(range(...))`.

If you need both the index and the element, loop over `range(len(...))`:

````python
print(list(range(2, 6)))

for i in range(len(names)):
    print(i, names[i], masses[i])
````

### Physics: the energy levels of hydrogen

**Physics recap.** In the Bohr model of hydrogen (which quantum mechanics confirms), the electron can only have certain discrete energies, labeled by a positive integer $n = 1, 2, 3, \ldots$ called the **principal quantum number**:
$$E_n = -\frac{13.6\ \text{eV}}{n^2}.$$
The energies are negative because the electron is *bound*: you would need to add energy $|E_n|$ to free it. $n = 1$ is the **ground state**, $E_1 = -13.6$ eV, and the levels crowd together toward $E = 0$ as $n$ grows. (The precise constant is the Rydberg energy, 13.6057 eV.)

Let us print a table of the first six levels.

````python
E_R = 13.6057     # Rydberg energy, eV

for n in range(1, 7):
    E_n = -E_R / n**2
    print(f"n = {n}:  E = {E_n:8.3f} eV")
````

The format spec `8.3f` makes every number take up 8 characters with 3 decimals, so the columns line up.

We can also store the energies in a list for later use, by starting with an empty list and appending:

````python
energies = []
for n in range(1, 7):
    energies.append(-E_R / n**2)

print(energies)
print(f"Energy needed to go from n=1 to n=2: {energies[1] - energies[0]:.3f} eV")
````

**Off-by-one pitfall:** in the list `energies`, the level $n = 1$ is stored at index 0, $n = 2$ at index 1, and so on. Mixing up "the $n$-th level" and "index $n$" is one of the most common bugs in all of programming. It has a name: an **off-by-one error**. Always ask yourself what the first and last values of your loop variable are.

## 3. Making decisions: `if`, `elif`, `else`

### Comparison operators

These compare two values and give a `bool` (`True` or `False`):

| Operator | Meaning |
|---|---|
| `==` | equal to (two equals signs!) |
| `!=` | not equal to |
| `<`, `>` | less than, greater than |
| `<=`, `>=` | less than or equal, greater than or equal |

### Logical operators

Combine conditions with `and`, `or`, and `not`:

- `a and b` is `True` only if both are `True`;
- `a or b` is `True` if at least one is `True`;
- `not a` flips `True` and `False`.

````python
wavelength = 500   # nm

print(wavelength > 400)
print(wavelength > 400 and wavelength < 700)
print(400 < wavelength < 700)     # Python allows chained comparisons, like math
print(wavelength < 400 or wavelength > 700)
print(not wavelength == 500)
````

### `if` statements

An `if` statement runs a block of code only when a condition is `True`. It uses the same colon-and-indentation structure as a loop:

```python
if condition:
    # runs only if condition is True
elif other_condition:
    # runs if the first condition was False and this one is True
else:
    # runs if none of the conditions above were True
```

You can have any number of `elif` ("else if") branches, and the `else` is optional. Python checks the conditions **in order** and runs **only the first** branch whose condition is true.

**Physics example:** the electromagnetic spectrum. Light of all wavelengths is the same phenomenon, but we give different ranges different names. Roughly: below 10 nm is X-ray (or gamma), 10 to 380 nm ultraviolet, 380 to 750 nm visible, 750 nm to 1 mm infrared, and longer wavelengths microwave and radio. (The boundaries are conventions, not sharp physics.)

````python
wavelength = 656.5   # nm

if wavelength < 10:
    band = "X-ray or gamma ray"
elif wavelength < 380:
    band = "ultraviolet"
elif wavelength <= 750:
    band = "visible"
elif wavelength < 1e6:
    band = "infrared"
else:
    band = "microwave or radio"

print(f"{wavelength} nm is {band}")
````

Because the conditions are checked in order, the `elif wavelength < 380` branch only runs if the first test failed, so we already know `wavelength >= 10` there. Change `wavelength` to a few different values and rerun the cell.

`if` statements can go inside loops (notice the two levels of indentation):

````python
for n in range(1, 11):
    if n % 2 == 0:
        print(n, "is even")
    else:
        print(n, "is odd")
````

## 4. `while` loops, `break`, and `continue`

A `for` loop runs a known number of times. A **`while` loop** keeps going **as long as a condition is true**, which is what you want when you do not know in advance how many steps are needed:

```python
while condition:
    # body: repeats while condition is True
```

**Example: radioactive decay.** A radioactive sample loses half of its remaining atoms every **half-life**. How many half-lives until less than 1% is left?

````python
fraction = 1.0       # fraction of the sample remaining
half_lives = 0

while fraction >= 0.01:
    fraction = fraction / 2
    half_lives += 1

print(f"After {half_lives} half-lives, {100 * fraction:.2f}% remains")
````

**Warning: infinite loops.** If the condition never becomes `False`, a `while` loop runs forever. For example, if you forgot the line `fraction = fraction / 2`, nothing would ever change. If a cell runs much longer than you expect (the spinning icon next to it never stops), click the stop button next to the cell, or use **Runtime → Interrupt execution**.

### `break` and `continue`

- `break` exits the loop immediately.
- `continue` skips the rest of the body and goes on to the next repetition.

Example: find the first hydrogen level $n$ whose energy is within 0.1 eV of zero (that is, an electron that is barely bound). We do not know which $n$ that will be, so we loop "forever" with `while True` and `break` when we find it.

````python
n = 1
while True:
    E_n = -E_R / n**2
    if abs(E_n) < 0.1:
        break            # leave the loop
    n += 1

print(f"First level with |E| < 0.1 eV: n = {n}, E = {E_n:.4f} eV")
````

````python
# continue: print only the odd levels
for n in range(1, 8):
    if n % 2 == 0:
        continue         # skip the rest of the body for even n
    print(n, -E_R / n**2)
````

## 5. Functions

You have already *used* many functions: `print`, `len`, `type`, `math.sqrt`. Now you will write your own. A **function** is a named, reusable block of code that takes some inputs (called **arguments**), does a calculation, and gives back (**returns**) a result.

Why bother? Functions let you write a calculation **once**, give it a clear name, test it, and then use it everywhere. If you find a bug, you fix it in one place.

Here is a function for the Lorentz factor $\gamma = 1/\sqrt{1-\beta^2}$:

````python
def lorentz_gamma(beta):
    """Return the Lorentz factor gamma for a speed beta = v/c."""
    gamma = 1 / math.sqrt(1 - beta**2)
    return gamma
````

Piece by piece:

- `def` starts a **function definition**, followed by the function's name, `lorentz_gamma`. Naming rules are the same as for variables.
- `(beta)` lists the **arguments** (inputs). Inside the function, `beta` is a variable holding whatever value the caller passed in.
- The line ends with a colon, and the **body** is indented, just like loops and `if`.
- The text in triple quotes `"""..."""` right after the `def` line is the **docstring**: a short description of what the function does. It is optional, but always write one. (You will see in Notebook 03 that Python's `help()` displays it.)
- `return gamma` ends the function and sends the value of `gamma` back to whoever called it.

Running the cell above **defines** the function but does not compute anything yet; there is no output. To use the function, you **call** it:

````python
print(lorentz_gamma(0.5))
print(lorentz_gamma(0.99))

g = lorentz_gamma(0.9)        # store the returned value in a variable
print(f"gamma at beta = 0.9: {g:.4f}")
````

### Several arguments, and default values

Functions can take several arguments, separated by commas. You can also give an argument a **default value** with `=` in the definition; the caller may then leave it out.

**Physics recap:** an electron in hydrogen jumps from a higher level $n_i$ (initial) to a lower level $n_f$ (final) by emitting a photon that carries away the energy difference. Using $E = hc/\lambda$ and $E_n \propto -1/n^2$, the photon's wavelength obeys the **Rydberg formula**:
$$\frac{1}{\lambda} = R_H \left( \frac{1}{n_f^2} - \frac{1}{n_i^2} \right), \qquad n_i > n_f,$$
with the Rydberg constant for hydrogen $R_H = 1.09678\times10^{7}\ \text{m}^{-1}$. Transitions that end on $n_f = 1$ form the **Lyman series** (ultraviolet), those ending on $n_f = 2$ the **Balmer series** (mostly visible), and those ending on $n_f = 3$ the **Paschen series** (infrared). These lines are the "fingerprint" by which astronomers identify hydrogen in stars.

````python
R_H = 1.09678e7     # Rydberg constant for hydrogen, 1/m

def hydrogen_wavelength(n_i, n_f=2):
    """Return the vacuum wavelength (in nm) of the photon emitted when a hydrogen
    electron drops from level n_i to level n_f (default n_f = 2, the Balmer series)."""
    inverse_lambda = R_H * (1 / n_f**2 - 1 / n_i**2)    # in 1/m
    return 1e9 / inverse_lambda                         # convert m to nm
````

````python
print(hydrogen_wavelength(3))          # n_f takes its default value 2: H-alpha
print(hydrogen_wavelength(2, 1))       # Lyman-alpha: 2 -> 1
print(hydrogen_wavelength(n_i=2, n_f=1))   # same, with the arguments named explicitly
````

The red H-alpha line at about 656 nm is the strongest visible hydrogen line, and the reason glowing hydrogen nebulae look pink-red in photographs. (These are *vacuum* wavelengths; tables often list wavelengths in air, which are about 0.03% shorter.)

Passing arguments by name, as in `n_i=2, n_f=1`, is called using **keyword arguments**. It makes calls easier to read and lets you give them in any order.

Now, with a function and a loop, we can print the whole Balmer series in a few lines:

````python
print("Balmer series (n_f = 2)")
for n_i in range(3, 9):
    lam = hydrogen_wavelength(n_i)
    print(f"  {n_i} -> 2:  {lam:7.1f} nm")
````

As $n_i \to \infty$ the lines crowd together toward the **series limit**, $\lambda = 4/R_H \approx 364.7$ nm.

### Scope: variables inside functions are local

Variables created inside a function (including its arguments) exist **only while the function runs**. They are called **local variables**, and the region of code where a name is visible is called its **scope**. A local variable does not affect a variable with the same name outside the function:

````python
gamma = 100.0                 # a variable outside any function ("global")

value = lorentz_gamma(0.6)    # inside, the function makes its own local "gamma"

print("returned value:", value)
print("global gamma is unchanged:", gamma)
````

This is a good thing: you can use a function without worrying about which names it uses internally. The only way information should get out of a function is through `return`.

A function *can* read a global variable that it never assigns to, as `hydrogen_wavelength` reads `R_H` above. That is convenient for true constants, but for anything that might change it is better to pass the value in as an argument, so the function's behavior depends only on its inputs.

**Pitfall: forgetting `return`.** If a function has no `return` statement, it returns the special value `None` ("nothing"). If you see `None` printed where you expected a number, check for a missing `return`. Also note that printing a value inside a function is not the same as returning it.

````python
def kinetic_energy_no_return(m, v):
    K = 0.5 * m * v**2          # computed... but never returned

print(kinetic_energy_no_return(2.0, 3.0))
````

## 6. A first simulation: the falling ball

You know the answer for a ball dropped from height $h$ with no air resistance: it hits the ground at time $t = \sqrt{2h/g}$. But most real problems (air resistance, planets pulling on each other, electrons in a potential) have no such formula. Instead, physicists **simulate**: they chop time into small steps $\Delta t$ and use the laws of physics to advance the system one step at a time.

**The idea.** Over a short time $\Delta t$, the velocity barely changes, so the position changes by approximately (velocity) $\times\ \Delta t$. Likewise, the velocity changes by (acceleration) $\times\ \Delta t$. For a falling ball, the acceleration is $-g$ (downward). So, starting from the current position $y$ and velocity $v$:
$$y_{\text{new}} = y + v\,\Delta t, \qquad v_{\text{new}} = v - g\,\Delta t, \qquad t_{\text{new}} = t + \Delta t.$$
Repeat until the ball reaches the ground. This simple recipe is called **Euler's method**. You do not need to know anything about differential equations to use it, and it is the starting point for Notebook 10, where we make it much more accurate.

Since we do not know in advance how many steps it will take, a `while` loop is the natural choice.

````python
g = 9.81       # gravitational acceleration, m/s^2
h = 20.0       # initial height, m
dt = 0.01      # time step, s

# initial conditions
t = 0.0
y = h
v = 0.0

while y > 0:
    y_new = y + v * dt     # use the OLD velocity to update the position
    v = v - g * dt
    y = y_new
    t += dt

print(f"Simulated: ball hits the ground at t = {t:.3f} s with v = {v:.3f} m/s")
print(f"Exact:     t = {math.sqrt(2 * h / g):.3f} s,  v = {-math.sqrt(2 * g * h):.3f} m/s")
````

Close, but not exact. There are two sources of error: Euler's method is only approximate, and the loop stops at the first time step *after* the ball crosses $y = 0$, so the landing time can be off by up to one $\Delta t$.

Let us package the simulation as a function so we can easily try different time steps, and record the trajectory in lists so we can plot it.

````python
def drop_ball(h, dt, g=9.81):
    """Simulate a ball dropped from rest at height h (m) with Euler's method,
    using time step dt (s). Return lists of times (s) and heights (m)."""
    t, y, v = 0.0, h, 0.0
    times = [t]
    heights = [y]
    while y > 0:
        y, v = y + v * dt, v - g * dt    # update both at once, using the old values
        t += dt
        times.append(t)
        heights.append(y)
    return times, heights
````

Two new things here: the line `y, v = y + v * dt, v - g * dt` computes *both* right-hand sides with the old values first, then assigns them, so we do not need the temporary `y_new`. And `return times, heights` returns two values at once; the caller can unpack them into two variables. (Notebook 03 explains that this is a *tuple*.)

````python
exact = math.sqrt(2 * 20.0 / 9.81)

for dt in [0.1, 0.01, 0.001, 0.0001]:
    times, heights = drop_ball(20.0, dt)
    t_land = times[-1]          # the last time in the list
    print(f"dt = {dt:7.4f} s:  {len(times):6d} steps,  t_land = {t_land:.4f} s,"
          f"  error = {t_land - exact:+.4f} s")
````

Smaller time steps give a more accurate answer, at the price of more steps (more computer time). Making the step 10 times smaller makes the error roughly 10 times smaller. This trade-off between accuracy and cost is a theme that runs through the whole course.

(Notice that a long line of code can be split inside parentheses: two f-strings written next to each other are joined into one.)

Here is the simulated trajectory together with the exact parabola $y = h - \frac{1}{2}gt^2$. Do not worry about the plotting commands yet; Notebook 05 covers them.

````python
import matplotlib.pyplot as plt

times, heights = drop_ball(20.0, 0.1)          # a large step so the difference is visible
exact_heights = [20.0 - 0.5 * 9.81 * t**2 for t in times]   # see the note below

plt.plot(times, heights, "o", label="Euler, dt = 0.1 s")
plt.plot(times, exact_heights, label="exact")
plt.xlabel("time (s)")
plt.ylabel("height (m)")
plt.legend()
plt.show()
````

The line `exact_heights = [... for t in times]` is a **list comprehension**, a compact way to build a list with a loop in one line. It does the same as:

```python
exact_heights = []
for t in times:
    exact_heights.append(20.0 - 0.5 * 9.81 * t**2)
```

You do not need to use list comprehensions yourself, but you will see them in other people's code.

## 7. Reading error messages

Everyone's code has errors, all the time, including professionals'. The skill to learn is not avoiding errors but **fixing them quickly**, and that starts with reading the error message. When something goes wrong, Python prints a **traceback** like this one:

```
---------------------------------------------------------------------------
ZeroDivisionError                         Traceback (most recent call last)
<ipython-input-12-3a7f>  in <cell line: 0>()
      1 for b in [0.5, 0.9, 1.0]:
----> 2     print(lorentz_gamma(b))

<ipython-input-7-9c2e> in lorentz_gamma(beta)
      1 def lorentz_gamma(beta):
      2     """Return the Lorentz factor gamma for a speed beta = v/c."""
----> 3     gamma = 1 / math.sqrt(1 - beta**2)
      4     return gamma

ZeroDivisionError: float division by zero
```

How to read it:

1. **Start at the bottom.** The last line gives the **type of error** (`ZeroDivisionError`) and a short message (`float division by zero`).
2. **Then look at the arrows `---->`**, which point at the lines where things went wrong. The traceback goes from your cell (top) down into the function where the error actually happened (bottom). "Most recent call last" means the deepest point is at the bottom.
3. **Think about the values.** Here the loop called `lorentz_gamma(1.0)`, and $1 - 1^2 = 0$. The physics agrees: $\gamma$ is infinite at $v = c$.

### Catching errors with `try` / `except`

Sometimes you expect that an error might happen and want to handle it instead of crashing. Python's `try`/`except` does that: it runs the `try` block, and if an error of the named type occurs, it jumps to the `except` block. Writing `except ZeroDivisionError as err:` stores the error in the variable `err` so we can print its message.

Below, we use `try`/`except` to show you several common errors **without stopping the notebook**.

### A debugging checklist

When your code fails or gives a strange answer:

1. Read the **last line** of the error message, then find the line marked with the arrow.
2. Print the values of the variables involved just before that line. Are they what you expect?
3. Check the usual suspects: typos in names, missing colons or parentheses, wrong indentation, off-by-one ranges, `=` vs `==`, a cell you forgot to run (use **Restart session and run all**).
4. If there is **no** error but the answer is wrong, test your code on a case where you know the answer (like comparing `drop_ball` with the exact formula). This is called **validation**, and it is an essential part of computational physics.
5. Still stuck? Copy the last line of the error message into a search engine. You are almost never the first person to see it.

## 8. Putting it together: a table of $\gamma$ versus $\beta$

Here is a short program that combines a function, a list, a loop, an `if`, and formatted printing. It prints a table of the Lorentz factor and of the relativistic kinetic energy of an electron, $K = (\gamma - 1)\,m_e c^2$ with $m_e c^2 = 0.511$ MeV, for a range of speeds, and flags the speeds where Newtonian kinetic energy $\frac{1}{2}mv^2 = \frac{1}{2}\beta^2 mc^2$ would be off by more than 10%.

````python
mc2 = 0.511     # electron rest energy, MeV
betas = [0.01, 0.1, 0.3, 0.5, 0.7, 0.9, 0.99, 0.999]

print(f"{'beta':>6} {'gamma':>9} {'K (MeV)':>10} {'K_Newton (MeV)':>15}")
for beta in betas:
    gamma = lorentz_gamma(beta)
    K = (gamma - 1) * mc2
    K_newton = 0.5 * beta**2 * mc2
    flag = ""
    if abs(K_newton - K) / K > 0.10:
        flag = "  <-- Newton off by > 10%"
    print(f"{beta:6.3f} {gamma:9.3f} {K:10.4f} {K_newton:15.4f}{flag}")
````

(`{'beta':>6}` right-aligns the text `beta` in a 6-character column, so the headings line up with the numbers.)

At $\beta = 0.3$ the Newtonian formula is already off by about 7%; by $\beta = 0.5$ it is off by more than 10%, and at $\beta = 0.999$ it is hopeless.

## Summary

- **Lists**: `x = [a, b, c]`; index from 0: `x[0]`, last `x[-1]`; slice `x[start:stop]` (stop not included); `len(x)`; `x.append(value)`; start empty with `[]`.
- **`for` loops**: `for item in my_list:` or `for i in range(n):`. `range(start, stop, step)` stops *before* `stop`. The loop body is the indented block after the colon.
- **Accumulate** with `total = 0` before the loop and `total += ...` inside it; build lists with `.append`.
- **Conditionals**: `if ...: / elif ...: / else:`; comparisons `== != < > <= >=`; logic `and or not`. The first true branch wins.
- **`while condition:`** repeats until the condition is false; watch for infinite loops. `break` exits a loop, `continue` skips to the next pass.
- **Functions**: `def name(arg1, arg2=default):`, a `"""docstring"""`, and `return value`. Variables inside are **local** (scope). No `return` means the function returns `None`.
- **Errors**: read the traceback from the **bottom**; follow the `---->` arrows. Common types: `NameError`, `TypeError`, `IndexError`, `ValueError`, `ZeroDivisionError`, `SyntaxError`, `IndentationError`. `try: ... except SomeError as err: ...` handles an expected error.
- **Off-by-one errors** are everywhere: check the first and last values of every loop.
- **Simulation by time stepping** (Euler's method): $y \to y + v\,\Delta t$, $v \to v + a\,\Delta t$, $t \to t + \Delta t$. Smaller $\Delta t$ is more accurate but takes more steps. Always **validate** against a known answer.
- Physics: $E_n = -13.6\ \text{eV}/n^2$; Rydberg formula $1/\lambda = R_H(1/n_f^2 - 1/n_i^2)$; Lyman ($n_f=1$), Balmer ($n_f=2$), Paschen ($n_f=3$).
