## 2. Variables and assignment

A **variable** is a name that refers to a value. You create one with the **assignment operator** `=`:

```python
name = value
```

**Important:** in Python, `=` does *not* mean "equals" in the mathematical sense. It means "compute the value on the right, then store it under the name on the left". It is an instruction, not a statement of fact. Read `x = 5` as "x gets 5" or "set x to 5".

````python
mass = 2.0          # kg
velocity = 3.0      # m/s

kinetic_energy = 0.5 * mass * velocity**2   # (1/2) m v^2, in joules
print(kinetic_energy)
````

Because `=` is an instruction, something like `x = x + 1` makes perfect sense in Python, even though it is nonsense as algebra. It means "take the current value of `x`, add 1, and store the result back in `x`".

````python
x = 10
print("before:", x)
x = x + 1
print("after: ", x)
````

A variable can be **reassigned** to a new value at any time; the old value is simply forgotten. Python always runs code **top to bottom**, one line at a time, so a variable must be assigned *before* the line that uses it.

You can also assign several variables in one line, which is handy for related quantities:

````python
x0, v0 = 0.0, 5.0     # initial position (m) and velocity (m/s)
print(x0, v0)
````

### Naming rules

Variable names in Python:

- may contain letters, digits, and underscores `_`;
- **cannot start with a digit** (`2x` is not allowed, `x2` is fine);
- cannot contain spaces or symbols like `-`, `.`, `$` (`kinetic-energy` would be read as `kinetic minus energy`);
- are **case sensitive**: `Energy`, `energy`, and `ENERGY` are three different variables;
- cannot be one of Python's reserved **keywords**, such as `if`, `for`, `while`, `def`, `return`, `import`, `True`, `False`, `None`, `and`, `or`, `not`, `in`, `class`, and `lambda`.

That last one bites physicists: you might want to call a wavelength `lambda`, but it is a keyword. Use `lam` or `wavelength` instead.

The usual Python style is lowercase words separated by underscores, like `photon_energy` or `electron_mass`. **Choose meaningful names.** `v` is fine for a velocity in a short formula, but `electron_speed` is much clearer than `es` or `x7` in a longer calculation.

Also avoid using names of built-in functions such as `print`, `type`, `sum`, `min`, `max`, or `list` for your variables. Python will let you do it, but then the original function stops working until you restart the session.

## 3. Basic types: `int`, `float`, `str`, `bool`

Every value in Python has a **type**, which tells Python what kind of thing it is and what you can do with it. The four basic types are:

| Type | Meaning | Examples |
|---|---|---|
| `int` | integer (whole number) | `3`, `-7`, `0`, `1000000` |
| `float` | "floating-point" number (has a decimal point) | `3.0`, `-0.5`, `2.998e8` |
| `str` | string (text) | `"electron"`, `'eV'`, `"3.0"` |
| `bool` | Boolean (true or false) | `True`, `False` |

You can ask Python for the type of any value with the built-in function `type()`:

````python
print(type(3))
print(type(3.0))
print(type("3.0"))
print(type(True))
````

Notice that `3`, `3.0`, and `"3.0"` are three different things: an integer, a float, and a piece of text that happens to look like a number. You cannot do arithmetic with text.

`bool` values usually come from **comparisons**, which you will use a lot in the next notebook. The comparison "is equal to" is written with a *double* equals sign `==`, to distinguish it from assignment `=`.

````python
speed = 2.5e8        # m/s
c = 299792458        # speed of light, m/s

print(speed < c)     # is the speed less than c?
print(speed == c)    # is it exactly equal to c?
````

### Converting between types

You can convert a value to another type by "calling" the type name like a function: `int(...)`, `float(...)`, `str(...)`, `bool(...)`.

````python
print(float(7))         # int -> float
print(int(3.99))        # float -> int: chops off the decimals (it does NOT round!)
print(round(3.99))      # use round() if you want rounding
print(float("6.626e-34"))   # text -> number
print(str(42) + " eV")      # number -> text, so it can be joined to other text
````

Converting text to a number is useful because data from files, and from the keyboard, arrives as text. (There is a function `input()` that asks the user to type something, and it always returns a string. We do not use it in these notebooks, because a notebook that stops to wait for typing cannot be run with "Run all".)

**A common error:** trying to combine a number and a string with `+`, as in `"Energy: " + 5`. Python does not guess what you mean, and stops with an error message like:

```
TypeError: can only concatenate str (not "int") to str
```

"Concatenate" means "join together". Python is saying: you can join a string to a string, but not to an integer. The fix is `"Energy: " + str(5)`, or better, an f-string (Section 6).

An error message is not a disaster. It is Python telling you, as precisely as it can, what went wrong. The **last line** of the message is usually the most useful: it gives the *type* of error (here `TypeError`) and a description. You will learn to read full error messages in Notebook 02.

## 4. Arithmetic

Python's arithmetic operators:

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+` | add | `7 + 2` | `9` |
| `-` | subtract | `7 - 2` | `5` |
| `*` | multiply | `7 * 2` | `14` |
| `/` | divide | `7 / 2` | `3.5` |
| `//` | integer ("floor") division | `7 // 2` | `3` |
| `%` | remainder ("modulo") | `7 % 2` | `1` |
| `**` | power | `7 ** 2` | `49` |

Two warnings for people used to other notation: powers use `**`, **not** `^` (in Python `^` means something completely different, and will silently give a wrong answer), and multiplication always needs an explicit `*`: write `2 * x`, never `2x`.

````python
print(7 + 2, 7 - 2, 7 * 2)
print(7 / 2)
print(7 // 2, 7 % 2)
print(7 ** 2)
print(2 ** 0.5)     # a power of 0.5 is a square root
````

### Integer versus float division

`/` is "true division" and **always** gives a float, even when the answer is a whole number: `6 / 3` is `2.0`, not `2`.

`//` divides and then rounds **down** to a whole number, and `%` gives what is left over. Together they answer questions like "how many whole hours and leftover minutes are in 200 minutes?":

````python
minutes = 200
hours = minutes // 60       # whole hours
leftover = minutes % 60     # remaining minutes
print(hours, "hours and", leftover, "minutes")

print(6 / 3)                # true division always gives a float
````

`%` is also handy to test whether a number is even (`n % 2` is `0`) or to make things repeat periodically, like wrapping an angle back into the range 0 to 360 degrees: `370 % 360` is `10`.

Mixing an `int` and a `float` in a calculation gives a `float`.

### Order of operations

Python follows the usual mathematical order (sometimes remembered as PEMDAS):

1. parentheses `( )`
2. powers `**`
3. multiplication and division `*`, `/`, `//`, `%` (left to right)
4. addition and subtraction `+`, `-` (left to right)

When in doubt, **add parentheses**. They never hurt and they make your intention clear. Here is a classic trap: $\frac{1}{2m}$ is *not* `1/2*m`.

````python
m = 4.0
print(1 / 2 * m)       # Python reads this as (1/2) * m = 2.0
print(1 / (2 * m))     # this is 1/(2m) = 0.125
````

Another trap involves powers and minus signs: `-2**2` is `-4`, because the power is done before the minus sign. Write `(-2)**2` if you mean the square of $-2$.

````python
print(-2**2)
print((-2)**2)
````

## 5. Scientific notation

Physics involves very large and very small numbers. Python writes $1.602\times10^{-19}$ as `1.602e-19`. The `e` means "times ten to the power", and it has nothing to do with Euler's number $e$. There must be no spaces inside the number.

| Math | Python |
|---|---|
| $3.00\times10^{8}$ | `3.00e8` |
| $1.602\times10^{-19}$ | `1.602e-19` |
| $6.626\times10^{-34}$ | `6.626e-34` |
| $10^{-9}$ | `1e-9` |

**Pitfall:** `10e-9` means $10\times10^{-9} = 10^{-8}$, not $10^{-9}$. Write `1e-9` for one nano-unit.

Numbers written this way are always floats.

````python
e_charge = 1.602176634e-19     # elementary charge, C (exact in SI since 2019)
N_A = 6.02214076e23            # Avogadro's number, 1/mol (exact)

print(e_charge * N_A)          # charge of one mole of electrons, C (the Faraday constant)
````

The answer, about 96 485 C/mol, is the Faraday constant from chemistry. Python displays very large or small results in scientific notation automatically.

## 6. Printing nicely with f-strings

Raw results often show more digits than you want, and without units. An **f-string** (formatted string) fixes this. Put the letter `f` right before the opening quote, and put any variable or expression inside curly braces `{ }`. Python replaces the braces with the value:

````python
mass = 2.0
velocity = 3.0
K = 0.5 * mass * velocity**2

print(f"A {mass} kg object moving at {velocity} m/s has K = {K} J")
print(f"Twice that energy is {2 * K} J")     # any expression works inside the braces
````

### Format specifications

After the expression inside the braces, you can add a colon `:` followed by a **format specification** that controls how the number is shown:

| Spec | Meaning | `{1234.56789:spec}` gives |
|---|---|---|
| `.2f` | fixed-point, 2 digits after the decimal point | `1234.57` |
| `.0f` | fixed-point, no decimals | `1235` |
| `.3e` | scientific notation, 3 digits after the decimal point | `1.235e+03` |
| `.4g` | "general": 4 significant figures, picks f or e automatically | `1235` |
| `,.1f` | with comma separators | `1,234.6` |
| `10.2f` | at least 10 characters wide (handy for tables) | `   1234.57` |

For physics, `.3e` (or however many digits are justified) is especially useful. These specs change only how the number is *displayed*; the stored value keeps all its digits.

````python
h = 6.62607015e-34      # Planck's constant, J s (exact)

print(f"h = {h}")
print(f"h = {h:.3e} J s")
print(f"h = {h:.2e} J s")
print(f"pi to 4 decimals: {3.14159265:.4f}")
````

## 8. A warning about floating-point numbers

Computers store floats in binary (base 2) with a fixed number of digits, about 16 significant decimal digits. Just as $1/3 = 0.3333\ldots$ cannot be written exactly in decimal, numbers like $0.1$ cannot be written exactly in binary. So tiny **round-off errors** appear:

````python
print(0.1 + 0.2)
print(0.1 + 0.2 == 0.3)
````

This is not a bug in Python; every programming language behaves this way. The error is around $10^{-16}$ relative to the size of the number, which is far below anything you could measure. Two practical lessons:

1. **Never test floats for exact equality with `==`.** Instead check whether they are *close*, for example `abs(a - b) < 1e-12`, or use `math.isclose(a, b)` (see the next section).
2. Do not panic about the last digit or two of a printed float. Use format specs to show a sensible number of digits.

Notebook 06 explores floating-point error in depth; it turns out to matter a lot in numerical calculations.

````python
a = 0.1 + 0.2
print(abs(a - 0.3) < 1e-12)    # abs() gives the absolute value
````

## 9. The `math` module

Python by itself does not know about $\sqrt{x}$, $\sin x$, or $\pi$. These live in a **module** called `math`, which comes with Python. A module is a collection of extra tools; you load it with an `import` statement, usually at the top of a notebook. Then you use its contents with a dot: `math.sqrt`, `math.pi`, and so on. (Notebook 03 covers modules and imports in detail.)

````python
import math

print(math.pi)
print(math.sqrt(2))
print(math.sin(math.pi / 6))       # trig functions use RADIANS
print(math.exp(1))                 # e^1, Euler's number
print(math.log(math.e))            # natural logarithm (ln)
print(math.log10(1000))            # base-10 logarithm
````

Things to remember:

- Trig functions use **radians**. Convert degrees with `math.radians(30)` (and back with `math.degrees`).
- `math.log` is the **natural** log $\ln x$. Use `math.log10` for $\log_{10} x$.
- `math.isclose(a, b)` checks whether two floats are equal up to round-off.
- Did you notice that $\sin(\pi/6)$ came out as `0.49999999999999994` instead of `0.5`? That is round-off again: `math.pi` is only an approximation of $\pi$ to about 16 digits.

````python
print(math.sin(math.radians(30)))      # sin(30 degrees)
print(math.isclose(0.1 + 0.2, 0.3))
````

## 10. Physics: relativity with Python

Now put it all together. First, a quick refresher on the physics.

**Special relativity in two equations.** When an object moves at speed $v$, define $\beta = v/c$ (the speed as a fraction of the speed of light $c$) and the **Lorentz factor**
$$\gamma = \frac{1}{\sqrt{1 - \beta^2}}.$$
At everyday speeds $\beta \approx 0$ and $\gamma \approx 1$, and Newtonian physics works. As $v \to c$, $\gamma$ grows without limit. Moving clocks run slow by a factor $\gamma$ (time dilation), and the momentum and energy of a particle of mass $m$ become
$$p = \gamma m v, \qquad E = \gamma m c^2, \qquad K = E - mc^2 = (\gamma - 1)\, m c^2.$$
Here $E$ is the total energy, $mc^2$ is the **rest energy** (the energy the particle has just by existing), and $K$ is the kinetic energy. A useful check is the energy-momentum relation $E^2 = (pc)^2 + (mc^2)^2$.

We will compute these for an electron moving at 90% of the speed of light. The constants below are the CODATA recommended values (the speed of light, Planck's constant, and the elementary charge are exact by definition of the SI units). From Notebook 03 on, you will load such constants from a library instead of typing them.

````python
c = 299792458.0            # speed of light, m/s (exact)
m_e = 9.1093837e-31        # electron mass, kg (CODATA, rounded)

beta = 0.9
v = beta * c
gamma = 1 / math.sqrt(1 - beta**2)

print(f"v     = {v:.4e} m/s")
print(f"gamma = {gamma:.4f}")
````

````python
p = gamma * m_e * v              # relativistic momentum, kg m/s
p_classical = m_e * v            # Newtonian momentum, for comparison

E_rest = m_e * c**2              # rest energy, J
E = gamma * E_rest               # total energy, J
K = E - E_rest                   # kinetic energy, J

print(f"p (relativistic) = {p:.4e} kg m/s")
print(f"p (classical)    = {p_classical:.4e} kg m/s")
print(f"Rest energy      = {E_rest:.4e} J")
print(f"Total energy     = {E:.4e} J")
print(f"Kinetic energy   = {K:.4e} J")
````

At $\beta = 0.9$ the relativistic momentum is more than twice the Newtonian one, since $\gamma \approx 2.29$.

Let us check our numbers with the energy-momentum relation. The two sides should agree up to round-off, so we compare them with `math.isclose` rather than `==`:

````python
lhs = E**2
rhs = (p * c)**2 + E_rest**2
print(f"E^2             = {lhs:.6e} J^2")
print(f"(pc)^2+(mc^2)^2 = {rhs:.6e} J^2")
print("Equal (up to round-off)?", math.isclose(lhs, rhs))
````

### Electron-volts

Joules are awkward for single particles: the numbers are tiny. Atomic and particle physicists use the **electron-volt** (eV), the energy an electron gains when it is accelerated through a potential difference of 1 volt:
$$1\ \text{eV} = 1.602176634\times10^{-19}\ \text{J} \quad\text{(exact)}.$$
Larger units are keV ($10^3$ eV), MeV ($10^6$ eV), and GeV ($10^9$ eV).

To convert joules to eV, **divide** by the number of joules per eV; to go back, **multiply**. It helps to put the conversion factor in a well-named variable so you never get confused about which way to go.

````python
J_per_eV = 1.602176634e-19       # joules in one electron-volt (exact)

E_rest_MeV = E_rest / J_per_eV / 1e6
K_MeV = K / J_per_eV / 1e6

print(f"Electron rest energy: {E_rest_MeV:.4f} MeV")
print(f"Kinetic energy at beta = {beta}: {K_MeV:.4f} MeV")
````

You should recognize $m_e c^2 \approx 0.511$ MeV, a number worth memorizing.

Particle physicists also give momentum in MeV/$c$: compute $pc$ in MeV, and quote $p$ in units of MeV/$c$.

````python
pc_MeV = p * c / J_per_eV / 1e6
print(f"p = {pc_MeV:.4f} MeV/c")
````

### Photon energy from wavelength

Light comes in packets called **photons**. A photon of frequency $f$ carries energy $E = hf$, where $h$ is Planck's constant. Since light travels at $c$, its wavelength and frequency are related by $c = f\lambda$, so
$$E = hf = \frac{hc}{\lambda}.$$
Shorter wavelength means higher energy: blue photons carry more energy than red ones, and X-rays far more than either.

Let us compute the energy of a photon of green light, $\lambda = 532$ nm (the color of a common laser pointer).

````python
h = 6.62607015e-34        # Planck's constant, J s (exact)
lam = 532e-9              # wavelength in m (we cannot call it "lambda": that is a keyword!)

E_photon = h * c / lam            # J
E_photon_eV = E_photon / J_per_eV # eV

print(f"E = {E_photon:.3e} J = {E_photon_eV:.3f} eV")
````

About 2.3 eV. Visible photons have energies of a few eV, which is why the electron-volt is so convenient for atomic physics.

A handy shortcut: $hc \approx 1240\ \text{eV nm}$, so $E\,[\text{eV}] \approx 1240 / \lambda\,[\text{nm}]$. Let us check that shortcut with Python:

````python
hc_eV_nm = h * c / J_per_eV / 1e-9     # hc in eV*nm
print(f"hc = {hc_eV_nm:.2f} eV nm")
print(f"1240 / 532 = {1240 / 532:.3f} eV")
````
